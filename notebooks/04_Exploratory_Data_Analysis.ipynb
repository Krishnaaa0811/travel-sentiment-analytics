```python
import re

def extract_city(address):
    match = re.search(r'\b\d{4,5}\s+([A-Za-zÀ-ÿ\s\-]+?)\s+(?:United Kingdom|Netherlands|France|Spain|Austria|Italy)$', address)
    
    if match:
        return match.group(1).strip()
    
    return None

df["Hotel_City"] = df["Hotel_Address"].apply(extract_city)

df["Hotel_City"].head(10)
```

```python
def extract_city(address):
    if address.endswith("United Kingdom"):
        address = address[:-len(" United Kingdom")].strip()
        
    elif address.endswith("Netherlands"):
        address = address[:-len(" Netherlands")].strip()
        
    elif address.endswith("France"):
        address = address[:-len(" France")].strip()
        
    elif address.endswith("Spain"):
        address = address[:-len(" Spain")].strip()
        
    elif address.endswith("Austria"):
        address = address[:-len(" Austria")].strip()
        
    elif address.endswith("Italy"):
        address = address[:-len(" Italy")].strip()

    parts = address.split()
    
    return parts[-1]


df["Hotel_City"] = df["Hotel_Address"].apply(extract_city)

df["Hotel_City"].head(10)
```

```python
df["Review_Word_Count"] = (
    df["Review_Text"]
    .fillna("")
    .str.split()
    .str.len()
)

df["Review_Word_Count"].describe()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df["Review_Word_Count"],
    bins=50,
    range=(0, 150)
)

plt.title("Distribution of Review Length")
plt.xlabel("Number of Words")
plt.ylabel("Number of Reviews")

plt.show()
```

```python
df["Score_Category"] = pd.cut(
    df["Reviewer_Score"],
    bins=[0, 5, 7, 10],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)

review_length_by_score = (
    df.groupby("Score_Category", observed=True)["Review_Word_Count"]
      .agg(["mean", "median", "count"])
)

review_length_by_score
```

```python
review_length_plot = review_length_by_score.reset_index()

plt.figure(figsize=(9, 6))

plt.bar(
    review_length_plot["Score_Category"],
    review_length_plot["mean"]
)

plt.title("Average Review Length by Score Category")
plt.xlabel("Score Category")
plt.ylabel("Average Review Length (Words)")

plt.show()
```
